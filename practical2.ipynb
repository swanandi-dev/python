```python
#right angle triangle pattern
for i in range(1,6):
  for j in range(1,i+1):
    print("*",end="")
  print()
```

```python
#inverted right angle triangle
for i in range(5,0,-1):
  for j in range(1,i+1):
    print("*",end="")
  print()
```

```python
#number triangle pattern
for i in range(1,6):
  for j in range(1,i+1):
    print(j,end="")
  print()
```

```python
#pyramid patter

row = 5
for i in range(1, row + 1):
    print(" " * (row - i) + "* " * i)
```

```python
# Alphabet pattern
ch = 65
for i in range (5):
  for j in range(i + 1):
    print(chr(ch),end="")
  ch += 1
  print()
```

```python
#flyod's triangle
num = 1
for i in range(1,6):
  for j in range(1,i+1):
    print(num,end=" ")
    num += 1
  print()
```

```python
#diamond pattern
row = 5
#upper part
for i in range(1,row + 1):
  print(" "*(row-i)+"* "*i)
# lower part
for i in range (row-1,0,-1):
  print(" "*(row - i)+"* "*i)
```
